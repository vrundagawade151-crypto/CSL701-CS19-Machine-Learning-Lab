```python
median_age = df['Age'].median()
df['Age'] = df['Age'].fillna(median_age)
print(f"Missing values in 'Age' after filling: {df['Age'].isnull().sum()}")
```

```python
# Task 10: Filter surviving passengers (Survived == 1)
survived_passengers_df = df[df['Survived'] == 1]
display(survived_passengers_df.head())
```

```python
# Task 11: Convert Fare column to a NumPy array
fare_array = df['Fare'].to_numpy()
print(f"Type of fare_array: {type(fare_array)}")
display(fare_array[:10]) # Display first 10 elements of the array
```

```python
# Task 12: Calculate maximum and average Fare using np.max() and np.mean()
max_fare = numpy.max(fare_array)
avg_fare = numpy.mean(fare_array)

print(f"Maximum Fare: {max_fare:.2f}")
print(f"Average Fare: {avg_fare:.2f}")
```

```python
# Task 13: Plot histogram of Age.
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df['Age'].dropna(), bins=30, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Task 14: Plot bar chart of Survival Count.

plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', data=df)
plt.title('Survival Count (0 = Perished, 1 = Survived)')
plt.xlabel('Survived')
plt.ylabel('Number of Passengers')
plt.xticks([0, 1], ['Perished', 'Survived'])
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Task 15: Plot scatter graph of Age vs Fare.

plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Fare', data=df)
plt.title('Age vs. Fare')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
